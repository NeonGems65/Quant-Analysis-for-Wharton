```python
def cleanDollarSigns(dictionary):
    for row in range(len(dictionary)):
        dictionary[row] = dictionary[row].replace("$","")
    return dictionary
    

def findAvg(dictionary):
    timesIter = 0
    sum = 0
    for row in range(len(dictionary)):
        sum += float(dictionary.iloc[row])
        timesIter += 1
    
    avg15YrPrice = sum/timesIter
    return avg15YrPrice

def statement15yrPast(statement):
    df = pd.DataFrame()
    for i in range(15):
        df = pd.concat([df, statement.iloc[:,i]], ignore_index=False)
    return df

def convertToNumeric(df):
    for i in range(len(df.columns)):
        df.iloc[:,i] = pd.to_numeric(df.iloc[:,i],errors="coerce")
    return df

```
