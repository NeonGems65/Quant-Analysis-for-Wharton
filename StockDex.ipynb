```python
def cleanDollarSigns(dictionary):
    for row in range(len(dictionary)):
        dictionary[row] = dictionary[row].replace("$","")
    return dictionary
    

def findAvg(dictionary):
    timesIter = 0
    sum = 0
    for row in range(len(dictionary)):
        sum += float(dictionary[row])
        timesIter += 1
    
    avg15YrPrice = sum/timesIter
    return avg15YrPrice
```

```python
past15YrPrices = cleanDollarSigns(past15YrPrices)
avg15YrPrice = findAvg(past15YrPrices)

pastYrPrices = cleanDollarSigns(pastYrPrices)
avgYrPrice = findAvg(pastYrPrices)

print(avg15YrPrice)
print(avgYrPrice)
```
